 Question 1 – Filter strings by length

```python
def filter_by_length(strings, k):
    return [s for s in strings if len(s) > k]

# Get input from the user
strings = input("Enter strings separated by spaces: ").split()
k = int(input("Enter the number K: "))

print(filter_by_length(strings, k))
```

Question 2 – `area()` for a circle or rectangle

```python
def area(shape, *dims):
    """Area of a circle (radius) or a rectangle (length, width)."""
    shape = shape.lower()
    if shape == "circle" and len(dims) == 1:
        return f"Circle Area: {3.14 * dims[0] ** 2:g}"
    if shape == "rectangle" and len(dims) == 2:
        return f"Rectangle Area: {dims[0] * dims[1]:g}"
    return "Invalid shape or wrong number of dimensions"


# Get input from the user
shape = input("Enter shape (circle/rectangle): ").strip().lower()

if shape == "circle":
    radius = float(input("Enter the radius: "))
    print(area(shape, radius))
elif shape == "rectangle":
    length = float(input("Enter the length: "))
    width = float(input("Enter the width: "))
    print(area(shape, length, width))
else:
    print("Invalid shape or wrong number of dimensions")
```

Question 3 – Split an email into username and domain

```python
def split_email(email):
    username, domain = email.split("@", 1)
    return f"Username: {username}, Domain: {domain}"


email = input("Enter an email address: ")   # e.g. support@google.com
print(split_email(email))
```

Question 4 – Sum only the numbers in a list

```python
import ast


def sum_numbers(items):
    total = 0
    for item in items:
        # bool is a subclass of int in Python, so exclude it explicitly
        if isinstance(item, (int, float)) and not isinstance(item, bool):
            total += item
    return total


# Get the whole list at once
items = ast.literal_eval(input("Enter the list: "))

print(f"Total Sum: {sum_numbers(items):g}")
```

 Question 5 – `Restaurant` class

```python
class Restaurant:
    def __init__(self):
        self.menu_items = {}         # {item_name: price}
        self.book_table = {}         # {table_number: customer_name}
        self.customer_orders = []    # [{"customer": ..., "items": [...], "total": ...}]

    def add_item_to_menu(self, name, price):
        self.menu_items[name] = price
        print(f"Added to menu: {name} - ${price:.2f}")

    def book_tables(self, table_number, customer_name):
        if table_number in self.book_table:
            print(f"Table {table_number} is already booked by {self.book_table[table_number]}.")
        else:
            self.book_table[table_number] = customer_name
            print(f"Table {table_number} booked for {customer_name}.")

    def customer_order(self, customer_name, items):
        unavailable = [item for item in items if item not in self.menu_items]
        if unavailable:
            print(f"Order failed. Not on the menu: {', '.join(unavailable)}")
            return
        total = sum(self.menu_items[item] for item in items)
        self.customer_orders.append(
            {"customer": customer_name, "items": items, "total": total}
        )
        print(f"Order taken for {customer_name}: {', '.join(items)} | Total: ${total:.2f}")

    def print_menu(self):
        print("\n--- MENU ---")
        for name, price in self.menu_items.items():
            print(f"{name:<15} ${price:.2f}")


# Perform the tasks
restaurant = Restaurant()

# 1. Add items to the menu
restaurant.add_item_to_menu("Burger", 8.50)
restaurant.add_item_to_menu("Pizza", 12.00)
restaurant.add_item_to_menu("Pasta", 10.25)
restaurant.add_item_to_menu("Salad", 6.75)

# 2. Make table reservations
restaurant.book_tables(1, "Ahmed")
restaurant.book_tables(2, "Sara")
restaurant.book_tables(1, "Omar")   # already booked

# 3. Take customer orders
restaurant.customer_order("Ahmed", ["Burger", "Pizza"])
restaurant.customer_order("Sara", ["Pasta", "Salad"])
restaurant.customer_order("Omar", ["Sushi"])   # not on the menu

# 4. Print the menu
restaurant.print_menu()
```

Question 6 – Bike Rental System

```python
class Bike:
    def __init__(self, name="", price=0.0):
        self._name = name
        self._price = price

    def get_name(self):
        return self._name

    def set_name(self, name):
        self._name = name

    def get_price(self):
        return self._price

    def set_price(self, price):
        self._price = price


class Customer:
    def __init__(self, name="", age=0):
        self._name = name
        self._age = age

    def get_name(self):
        return self._name

    def set_name(self, name):
        self._name = name

    def get_age(self):
        return self._age

    def set_age(self, age):
        self._age = age


class MainBikeRental:
    def __init__(self):
        self._bikes = []      # bikes currently in stock
        self._rentals = []    # [{"customer": Customer, "bikes": [Bike, ...]}]

    def get_bikes(self):
        return self._bikes

    def set_bikes(self, bikes):
        self._bikes = bikes

    def get_rentals(self):
        return self._rentals

    def set_rentals(self, rentals):
        self._rentals = rentals

    def addBike(self, bike):
        self._bikes.append(bike)

    def totalCost(self):
        return sum(bike.get_price() for bike in self._bikes)

    def requestBike(self, customer, num_of_bikes):
        if customer.get_age() <= 6:
            print(f"Sorry {customer.get_name()}, you must be older than 6 years to rent a bike.")
            return False
        if num_of_bikes <= 0:
            print("Please request at least one bike.")
            return False
        if num_of_bikes > len(self._bikes):
            print(f"Only {len(self._bikes)} bike(s) available. Please wait 10 mins.")
            return False

        # Deduct the bikes from the shop stock
        rented = [self._bikes.pop() for _ in range(num_of_bikes)]
        self._rentals.append({"customer": customer, "bikes": rented})
        print(f"{customer.get_name()} rented {num_of_bikes} bike(s). Stock left: {len(self._bikes)}")
        return True

    def returnBike(self, customer, rentalTime, num_of_bikes):
        for rental in self._rentals:
            if rental["customer"] is customer:
                if num_of_bikes > len(rental["bikes"]):
                    print(f"{customer.get_name()} only has {len(rental['bikes'])} bike(s) rented.")
                    return 0
                # Put the bikes back into stock
                for _ in range(num_of_bikes):
                    self._bikes.append(rental["bikes"].pop())
                if not rental["bikes"]:
                    self._rentals.remove(rental)

                bill = rentalTime * 40
                print(f"{customer.get_name()} returned {num_of_bikes} bike(s). "
                      f"Bill: {bill} | Stock now: {len(self._bikes)}")
                return bill

        print(f"No active rental found for {customer.get_name()}.")
        return 0


# ---------- Demo ----------
shop = MainBikeRental()

shop.addBike(Bike("Mountain", 500))
shop.addBike(Bike("Road", 400))
shop.addBike(Bike("City", 300))
shop.addBike(Bike("BMX", 350))

print("Total cost of bikes in store:", shop.totalCost())

ahmed = Customer("Ahmed", 25)
sara = Customer("Sara", 5)
omar = Customer("Omar", 18)

shop.requestBike(ahmed, 2)   # OK
shop.requestBike(sara, 1)    # too young
shop.requestBike(omar, 5)    # not enough stock

shop.returnBike(ahmed, 3, 2) # 3 hours -> 3 * 40 = 120
print("Total cost of bikes in store:", shop.totalCost())
```
